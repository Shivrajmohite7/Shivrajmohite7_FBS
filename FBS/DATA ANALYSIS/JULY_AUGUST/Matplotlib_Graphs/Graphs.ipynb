```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
month = ["Jan", "Feb", "Mar", "Apr", "May",
         "Jun", "Jul", "Aug", "Sep", "Oct",
         "Nov", "Dec"]

sales = [12000, 15000, 11000, 18000, 22000,
         19000, 25000, 28000, 23000, 30000,
         35000, 40000]

profit = [2000, 3000, 1500, 4000, 5000,
          3500, 6000, 7000, 5500, 8000,
          9000, 11000]

expenses = [8000, 9000, 7500, 11000, 13000,
            12000, 15000, 16000, 14000, 17000,
            19000, 21000]

customers = [100, 120, 90, 150, 180,
             160, 200, 220, 190, 250,
             280, 320]

products = ["Laptop", "Mobile", "Tablet", "Keyboard",
            "Mouse", "Monitor", "Laptop", "Mobile",
            "Tablet", "Keyboard", "Mouse", "Monitor"]

quantity = [15, 25, 12, 40, 60, 20,
            18, 30, 15, 50, 75, 25]

price = [50000, 20000, 15000, 1000, 500, 12000,
         52000, 21000, 16000, 1100, 550, 12500]

advertising = [1000, 1500, 800, 2000, 2500, 1800,
               3000, 3200, 2200, 3500, 4000, 4500]

rating = [3.5, 4.0, 3.8, 4.2, 4.5, 4.1,
          4.6, 4.3, 3.9, 4.7, 4.8, 4.9]
```

```python
plt.figure(figsize=(10,5))
plt.plot(month,sales,marker="o",color="pink",linewidth=2)
plt.title("MONTHLY SALES ANALYSIS",fontsize=20)
plt.xlabel("MONTHS",fontsize=15)
plt.ylabel("SALES",fontsize=15)
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend("Months")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 2. Bar Chart
# Beginner
# - X-axis: month
# - Y-axis: sales
# - Task: Compare monthly sales.
```

```python
plt.figure(figsize=(10,5))
plt.bar(month,sales,color=["red","green","blue","yellow","red","green","blue","yellow","red","green","blue","yellow"],label="Sales",width=0.5)
plt.title("MONTHLY SALES",fontsize=20)
plt.xlabel("MONTHS",fontsize=15)
plt.ylabel("SALES",fontsize=15)
plt.xticks(rotation=45)
plt.legend()
plt.grid(True,linestyle="--",alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
# 3. Horizontal Bar Chart
# - Y-axis: month
# - X-axis: profit
# - Task: Compare monthly profits horizontally.
```

```python
plt.figure(figsize=(10,5))
plt.barh(month,sales,color=["red","green","blue","yellow","red","green","blue","yellow","red","green","blue","yellow"],label="Sales",height=0.7)
plt.title("MONTHLY SALES",fontsize=20)
plt.xlabel("SALES",fontsize=15)
plt.ylabel("MONTH",fontsize=15)
plt.yticks(rotation=45)
plt.grid(True,linestyle="--",alpha=0.2)
plt.tight_layout()
plt.legend()
plt.show()
```

```python
# 4. Scatter Plot
# Intermediate
# - X-axis: advertising
# - Y-axis: sales
# - Task: Check whether higher advertising is associated with higher sales.
```

```python
plt.figure(figsize=(10,5))
plt.scatter(month,sales,color=["red","green","blue","yellow","red","green","blue","yellow","red","green","blue","yellow"],label="Sales")
plt.title("MONTHLY SALES")
plt.xlabel("MONTHS",fontsize=15)
plt.ylabel("SALES",fontsize=15)
plt.xticks(rotation=45)
plt.legend()
plt.grid(True,linestyle="--",alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# 5. Pie Chart
# - Labels: products
# - Values: quantity
# - Task: Show the percentage share of each product.
```

```python
plt.figure(figsize=(10,5))
plt.pie(sales,labels=month,autopct='%1.1f%%',colors=["red","green","blue","yellow","red","green","blue","yellow","red","green","blue","yellow"],pctdistance=0.65,
    labeldistance=1.1,)
plt.title("MONTHLY SALES")
plt.legend(loc="center left")
plt.tight_layout()
plt.show()
```
