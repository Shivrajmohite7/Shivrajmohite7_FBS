```python
movie_vs_tv=df.groupby(['release_year','type']).size().unstack().fillna(0)
fig,ax=plt.subplots(2,2,figsize=(12,10))
ax[0,0].plot(movie_vs_tv.index,movie_vs_tv['TV Show'],color='blue')
ax[0,0].set_title("TV Relased Per Year")
ax[0,0].set_xlabel("Year")
ax[0,0].set_ylabel("Number of Movies")

ax[0,0].plot(movie_vs_tv.index,movie_vs_tv['TV Show'],color='orange')
ax[0,0].set_title("TV Shows Relased Per Year")
ax[0,0].set_xlabel("Year")
ax[0,0].set_ylabel("Number of TV Shows")

ax[0,1].bar(movie_vs_tv.index, movie_vs_tv['Movie'], color='blue')

ax[0,1].set_title("Movies Released Per Year")
ax[0,1].set_xlabel("Year")
ax[0,1].set_ylabel("Number of Movies")

ax[1,0].bar(movie_vs_tv.index, movie_vs_tv['TV Show'], color='orange')

ax[1,0].set_title("TV Shows Released Per Year")
ax[1,0].set_xlabel("Year")
ax[1,0].set_ylabel("Number of TV Shows")

ax[1,1].set_title('Movies Realsed Per Year')
ax[1,1].plot(movie_vs_tv.index,movie_vs_tv['Movie'],color='blue')
ax[1,1].set_xlabel("Year")
ax[1,1].set_ylabel("Number of Movies")
plt.tight_layout()

plt.savefig('Subplot.png',dpi=100,bbox_inches='tight')
plt.show()
```
