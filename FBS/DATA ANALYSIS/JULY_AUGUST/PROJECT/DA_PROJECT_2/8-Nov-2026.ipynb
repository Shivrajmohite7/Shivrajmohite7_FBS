```python
# Identify products with high sales but low customer ratings.
Prod_Analysis=df.groupby('Product').agg( 
    Total_Sales=('Sales','sum'),
    Avg_Rating=('Customer_Rating','mean'))
High_Sales=Prod_Analysis['Total_Sales']>Prod_Analysis['Total_Sales'].mean()
Low_Rating=Prod_Analysis['Avg_Rating']<df['Customer_Rating'].mean()
Result=Prod_Analysis[High_Sales & Low_Rating]
print(Result)
```

```python
# Bar chart – Revenue by category.
Cat_Rev=df.groupby('Category')['Sales'].sum()
plt.figure(figsize=(8, 5))
plt.title('Revenue By Category')
plt.xlabel('Category')
plt.ylabel('Revenue')
plt.bar(Cat_Rev.index,Cat_Rev.values,color=['Red','Yellow','Blue','Brown'])
plt.show()
```

```python
# Horizontal bar chart – Top 10 products by sales.
Top_10_By_Sales=df.groupby('Product')['Sales'].sum()
Top_10_By_Sales=Top_10_By_Sales.sort_values(ascending=True).head(10)
plt.figure(figsize=(18,10))
plt.barh(Top_10_By_Sales.index,Top_10_By_Sales.values,color=['Red','Orange','Green','Blue','Black','Yellow','Pink','Purple','Brown','Red'])
plt.title("10 Products By Revenue")
plt.xlabel('Category')
plt.ylabel('Sales')
plt.show()
```

```python
# Pie chart – Order status distribution.
Order_Status=df.groupby('Order_Status')['Sales'].sum()
print(Order_Status)
```

```python
plt.pie(Order_Status,labels=['Delivered','Returned','Cancelled'],
        autopct="%1.1f%%",
        colors=['red','green','yellow'])
plt.title('Order_Status_Distribution')
plt.show()
```

```python
# Histogram – Distribution of order values.
x=df['Sales']
xx=list(x)
plt.hist(xx,color='Red',bins=3)
plt.title('Distribution of Order Values')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.show()
```

```python
# Scatter plot – Discount percentage vs total sales

DisvsSl = df.groupby('Discount_Percent')['Sales'].sum()

plt.figure(figsize=(8, 5))

plt.scatter(DisvsSl.index, DisvsSl.values, color='blue')

plt.title('Discount Percentage vs Total Sales')
plt.xlabel('Discount Percentage')
plt.ylabel('Total Sales')

plt.show()
```
