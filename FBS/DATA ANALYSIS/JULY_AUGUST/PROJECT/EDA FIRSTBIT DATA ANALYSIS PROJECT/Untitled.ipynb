```python
# def get_cate(hour):
#     if hour<6:
#         print("night")
#     elif hour<12:
#         print("morning")
#     elif hour<17:
#         print("afternoon")
#     elif pd.isna(hour):
#         print("Unknown")

# df['Time_Category']=df['Start_Hour'].apply(get_cate)
# df[['Time_Category','Start_Hour']]


def get_category(hour):
    if pd.isna(hour):
        return "Unknown"
    elif hour < 6:
        return "Night"
    elif hour < 12:
        return "Morning"
    elif hour < 17:
        return "Afternoon"
    else:
        return "Evening"

df['Time_Category'] = df['Start_Hour'].apply(get_category)
```
